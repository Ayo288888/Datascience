Concepts to know in git


1. Clone - allows y ou to copy an entire codebase 
2. commit -  commit is a permanent save point (snapshot) of your project's  files at a specific moment in time, stored safely in your local repository's history
3. Push - Uploads your saved changes from your computer to GitHub so others can see them.
4. Pull - Downloads new updates from GitHub and immediately blends them into your active files 
5. Staging -  A "waiting room" where you pick which modified files you want to include in your next save point
